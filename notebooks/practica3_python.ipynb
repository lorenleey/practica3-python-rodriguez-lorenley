## Ejercicio 2 — Inventario multialmacén con listas anidadas

```python
# Declaración de lista con sublistas
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

## Nombre del producto de la cuarta fila (doble indexación)
print("Producto de la cuarta fila:",inventario[3][1])

# Nueva lista donde cada elemento sea [producto, unidades * coste_unitario]
valor_stock = []

for fila in inventario:
    producto = fila[1]
    unidades = fila[2]
    coste_unitario = fila[3]

    valor = round((unidades*coste_unitario),2)
    valor_stock.append([producto, valor])

print("Valor del stock:")
print(valor_stock)

# Calcular la suma de valores del stock de TODOS los productos
valor_total = 0 

for fila in valor_stock:
    valor_total+= fila[1]

print("Valor total del stock:", valor_total)
# Slicing de sublista con solo filas de ALM-SUR 

alm_sur = inventario[2:4]
print("Inventario ALM-SUR:",alm_sur)

valor_sur = 0

for fila in alm_sur:
    valor_sur += round((fila[2]*fila[3]),2)

print("Valor total del stock de ALM-SUR:", valor_sur)
```

#### Información
El problema cuando los dos almacenes no estan contiguos es que si se desea unificar sus datos habría que hacer dos o más sublistas. Eso explica porque slicing no es una estrategia robusta de filtrado, ya que depende de la secuencia y orden de los elementos dentro de la lista.

## Ejercicio 5 — Funciones propias para el cálculo de importes

```python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """
    Calcula el importe neto de una línea de pedido
    aplicando un descuento porcentual.
    """
    bruto = unidades * precio_unitario
    neto = bruto * (1 - descuento_pct / 100)

    return round(neto, 2)


def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """
    Devuelve una tupla con importe bruto, ahorro e importe neto.
    """
    importe_bruto = unidades * precio_unitario
    ahorro = importe_bruto * descuento_pct / 100
    importe_neto = importe_bruto - ahorro

    return (
        round(importe_bruto, 2),
        round(ahorro, 2),
        round(importe_neto, 2)
    )

bruto, ahorro, neto = resumen_pedido(
    unidades=6,
    precio_unitario=40.90,
    descuento_pct=5
)

print("Bruto:", bruto)
print("Ahorro:", ahorro)
print("Neto:", neto)

# Documentación
help(calcular_importe)


# Calcular importe de pedidos
pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]

importes = []

for pedido in pedidos:
    unidades = pedido[0]
    precio = pedido[1]
    descuento = pedido[2]

    importe = calcular_importe(unidades, precio, descuento)
    importes.append(importe)

print("\nImportes:", importes)
print("Máximo:", max(importes))
print("Mínimo:", min(importes))
print("Ordenados:", sorted(importes))
print("Cantidad:", len(importes))
```

## Ejercicio 6 — Normalización de códigos SKU

```python
skus_crudos = [
    "  elc-0012-es ", "HOG-0045-ES", "dep-0003-pt  ",
    "ofi-0021-es", "  ELC-0012-ES", "hog-0099-fr ",
]

# Limpiar datos
skus_limpios = []

for sku in skus_crudos:
    limpio = sku.strip().upper()
    skus_limpios.append(limpio)
    
print("SKU limpios: ", skus_limpios)

# Contar apariciones
aparicionesElc = skus_limpios.count("ELC-0012-ES")
print("\nApariciones de ELC-0012-ES: ", aparicionesElc)

primeraAparicion = skus_limpios.index("OFI-0021-ES")
print("\nPrimera aparición de OFI-0021-ES: ", primeraAparicion)

# Separar SKU
partes_skus = []

for sku in skus_limpios:
    partes = sku.split("-")
    partes_skus.append(partes)

print("\nSKU separados:")
print(partes_skus)

# Filtrar SKU

skus_es = []

for sku in skus_limpios:
    partes = sku.split("-")
    
    if partes[2] == "ES":
        skus_es.append(sku.replace("-", "."))
        
print("\nSKU de España:", skus_es)

# Cadena única
cadena_skus = "| ".join(skus_es)

print("\nCadena final:", cadena_skus)
```

### Ejercicio 8 — Del cálculo con listas al cálculo vectorizado

```python
unidades = [3, 1, 5, 2, 4, 6, 2, 8, 3, 1]
precios  = [59.90, 279.00, 18.50, 149.00, 89.90, 27.50, 219.00, 14.90, 64.00, 239.00]

# Calcular 
print(unidades * precios)
```

Las listas de Python no permiten multiplicación elemento a elemento entre dos listas; para realizar esta operación se necesitan arrays NumPy. 

```python
# numpy arrays
import numpy as np

np_unidades = np.array(unidades)
np_precios = np.array(precios)

print("Unidades:")
print("dtype:", np_unidades.dtype)
print("shape:", np_unidades.shape)

print("\nPrecios:")
print("dtype:", np_precios.dtype)
print("shape:", np_precios.shape)

# Multiplicación elemento por elemento
importes = np_unidades * np_precios

print("Importes: ", importes)
print("Total: ", importes.sum(), "€")

# recargo 
importes_recargo = importes * 1.04

print("Importes con recargo:", importes_recargo)

# array mixto
mixto = np.array([1, "dos", 3.0, True])

print(mixto)
print("dtype:", mixto.dtype)
```

```python
# bucle con NumPy
def calcular_con_for():
    resultado = []

    for i in range(len(unidades)):
        resultado.append(unidades[i] * precios[i])

    return resultado
```

### Ejercicio 11 — Estadística descriptiva sobre telemetría simulada

```python
import numpy as np

# Generar dos arrays de 500 elementos
rng = np.random.default_rng(2024)

temperatura = rng.normal(68, 6.5, 500)
vibracion = np.abs(rng.normal(2.4, 0.8, 500))

# Calculos sobre temperatura
media = temperatura.mean()
mediana = np.median(temperatura)
desviacion = temperatura.std()
minimo = temperatura.min()
maximo = temperatura.max()

print("Media:", media)
print("Mediana:", mediana)
print("Desviación:", desviacion)
print("Mínimo:", minimo)
print("Máximo:", maximo)

# Calcular percentiles de vibración
percentiles = np.percentile(vibracion, [25, 50, 75, 95])
print("\nPercentiles vibración:", percentiles)

# Apilar arrays
telemetria = np.column_stack((temperatura, vibracion))
print("\nShape:", telemetria.shape)

# Calcular matriz de correlación
correlacion = np.corrcoef(temperatura, vibracion)
print("\nCorrelación:")
print(correlacion)

# Definir umbral
umbral = media + 2 * desviacion
superiores = (temperatura > umbral).sum()
porcentaje = round(superiores / len(temperatura) * 100, 2)

print("\nUmbral:", umbral)
print("Lecturas superiores:", superiores)
print("Porcentaje:", porcentaje, "%")
```

### Información
Fijar `default_rng(2024)` permite generar exactamente los mismos valores aleatorios en cada ejecución. Esto hace que otro analista pueda reproducir el mismo análisis y obtener los mismos resultados.

### Ejercicio 15 — Columnas calculadas, agregación y exportación

```python
# Crear columnas
ventas["importe_bruto"] = ventas["unidades"] * ventas["precio_unitario"]

ventas["importe_neto"] = (
    ventas["importe_bruto"] *
    (1 - ventas["descuento_pct"] / 100)
).round(2)


ventas["ticket_medio"] = ventas["importe_neto"] / ventas["unidades"]

# Agrupación
facturacion_region = (
    ventas.groupby("region")["importe_neto"]
    .sum()
    .sort_values(ascending=False)
)

print("Facturación por región:")
print(facturacion_region)


resumen_categoria_canal = (
    ventas.groupby(["categoria", "canal"])
    .agg({
        "importe_neto": "mean",
        "unidades": "sum"
    })
)

print("\nCategoría y canal:")
print(resumen_categoria_canal)

# Top 5 productos 
top_productos = (
    ventas.groupby("producto")["importe_neto"]
    .sum()
    .nlargest(5)
)

print("\nTop 5:")
print(top_productos)

# Rellenar valores nulos 
ventas["satisfaccion"] = ventas["satisfaccion"].fillna(
    ventas["satisfaccion"].median()
)

ventas["canal"] = ventas["canal"].fillna("Desconocido")

# Comprobación
print("\nFacturación total:", round(ventas["importe_neto"].sum(), 2), "€")

# Exportar a un .csv
facturacion_region.to_csv(
    "../outputs/facturacion_por_region.csv",
    header=["importe_neto"]
)

print("\nArchivo exportado.")
```

### Ejercicio 16 — Comparaciones y booleanos

```python

print(2 < 3)
print("Norte" == "norte")
print(True + True + False)
print(3 != 3.0)
print("Zapatilla" > "Mochila")


unidades = 6
precio = 149.0
descuento = 15

# Contrucción de condiciones
condicion1 = (unidades > 4) and (descuento >= 10)
print("\nCondición 1:", condicion1)


condicion2 = (precio < 50) or (precio > 200)
print("Condición 2:", condicion2)

# Negación
print("Negación:", not condicion1)


temp = np.array([64.2, 88.1, 71.0, 92.5, 69.8])
vib = np.array([1.9, 3.4, 2.1, 4.0, 2.2])

# 6
try:
    print(temp > 70 and vib > 3)
except ValueError as error:
    print("\nError:", error)

# Solución a comparativa 
resultado_logical = np.logical_and(temp > 70, vib > 3)
resultado_operador = (temp > 70) & (vib > 3)

print(resultado_logical)
print(resultado_operador)
print("Son iguales:", np.array_equal(resultado_logical, resultado_operador))
```

### Información
- `2 < 3` es `True` porque 2 es menor que 3.
- `"Norte" == "norte"` es `False` porque Python distingue mayúsculas y minúsculas.
- `True + True + False` devuelve 2 porque `True` vale 1 y `False` vale 0.
- `3 != 3.0` es `False` porque ambos representan el mismo valor numérico.
- `"Zapatilla" > "Mochila"` es `True` por el orden lexicográfico de las cadenas.

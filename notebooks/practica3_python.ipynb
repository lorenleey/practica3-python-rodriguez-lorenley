## Ejercicio 1 — Catálogo de productos como lista

```python
catalogo = [
    "Auriculares BT", 59.90,
    "Smartwatch S2", 149.00,
    "Tablet 10", 219.00,
    "Cargador USB-C", 18.50,
    "Cafetera Expres", 189.00,
    "Robot Aspirador", 279.00,
]

# 1. Tomando en cuenta que la lista tiene nombe y precio, asumimos que la mitad de elementos serán los productos
total_elementos = len(catalogo)
total_productos = total_elementos // 2

print("elementos:",total_elementos)
print("productos:",total_productos)

# slicing considerando que los nombres son los números impares, y precios los pares.
nombres = catalogo[::2]
precios = catalogo[1::2]

print("nombres:", nombres)
print("precios:", precios)

# Buscamos "Robot aspirador"
posicion_robot = catalogo.index("Robot Aspirador")
precio_robot = catalogo[posicion_robot + 1]

print("precio robot:", precio_robot)

#Tres ultimos nombres usando indíce negativo
ultimos_tres_nombres = nombres[-3::]
print(ultimos_tres_nombres)

# Precio medio de productos redondeando a dos deciamles unicamente
precio_medio = round(sum(precios) / len(precios),2)
print("precio medio:", precio_medio)
```

## Ejercicio 2 — Inventario multialmacén con listas anidadas

```python
# Declaración de lista con sublistas
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

## Nombre del producto de la cuarta fila (doble indexación)
print("Producto de la cuarta fila:",inventario[3][1])

# Nueva lista donde cada elemento sea [producto, unidades * coste_unitario]
valor_stock = []

for fila in inventario:
    producto = fila[1]
    unidades = fila[2]
    coste_unitario = fila[3]

    valor = round((unidades*coste_unitario),2)
    valor_stock.append([producto, valor])

print("Valor del stock:")
print(valor_stock)

# Calcular la suma de valores del stock de TODOS los productos
valor_total = 0 

for fila in valor_stock:
    valor_total+= fila[1]

print("Valor total del stock:", valor_total)
# Slicing de sublista con solo filas de ALM-SUR 

alm_sur = inventario[2:4]
print("Inventario ALM-SUR:",alm_sur)

valor_sur = 0

for fila in alm_sur:
    valor_sur += round((fila[2]*fila[3]),2)

print("Valor total del stock de ALM-SUR:", valor_sur)
```

#### Información
El problema cuando los dos almacenes no estan contiguos es que si se desea unificar sus datos habría que hacer dos o más sublistas. Eso explica porque slicing no es una estrategia robusta de filtrado, ya que depende de la secuencia y orden de los elementos dentro de la lista. 

## Ejercicio 3 — Manipulación destructiva y copias de listas

```python
precios_originales = [59.90, 149.00, 219.00, 18.50, 189.00, 279.00] 

# Precios rebajados con elementos modificados
precios_rebajados = precios_originales
precios_rebajados[0] = 49.90

print(precios_originales)
print(precios_rebajados)
```

```python
# Copia de una lista con list() y slicing [:]
precios_originales = [59.90, 149.00, 219.00, 18.50, 189.00, 279.00] 
precios_rebajados = list(precios_originales)
precios_rebajados_slc = precios_originales[:]

precios_rebajados[0] = 13.5
precios_rebajados_slc[2] = 100.90

print("Precios originales:",precios_originales)
print("Precios rebajados con list:",precios_rebajados)
print("Precios rebajados slicing:",precios_rebajados_slc)
```

```python
# Aplicar rebaja 10% a todos los elementos
precios_rebajas_10 = list(precios_originales)

for i in range(len(precios_rebajas_10)):
    precios_rebajas_10[i] = round((precios_rebajas_10[i] * 0.90),2)

# Añadir dos productos nuevos
precios_rebajas_10.extend([280.5, 345.2])
precios_rebajas_10.insert(2,560)

# Eliminar producto con del y pop()
del precios_rebajas_10[4]
descartado= precios_rebajas_10.pop()

print("Luego de eliminar y modificar:")
print(precios_rebajas_10)
print("Valor descartado:", descartado)

# Comprobar id 
print("\nid original:", id(precios_originales))
print("id precios list:", id(precios_rebajados))
print("id precios slicing:", id(precios_rebajados_slc))
print("id precios rebajados 10%:", id(precios_rebajas_10))

```

## Ejercicio 5 — Funciones propias para el cálculo de importes

```python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """
    Calcula el importe neto de una línea de pedido
    aplicando un descuento porcentual.
    """
    bruto = unidades * precio_unitario
    neto = bruto * (1 - descuento_pct / 100)

    return round(neto, 2)


def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """
    Devuelve una tupla con importe bruto, ahorro e importe neto.
    """
    importe_bruto = unidades * precio_unitario
    ahorro = importe_bruto * descuento_pct / 100
    importe_neto = importe_bruto - ahorro

    return (
        round(importe_bruto, 2),
        round(ahorro, 2),
        round(importe_neto, 2)
    )

bruto, ahorro, neto = resumen_pedido(
    unidades=6,
    precio_unitario=40.90,
    descuento_pct=5
)

print("Bruto:", bruto)
print("Ahorro:", ahorro)
print("Neto:", neto)

# Documentación
help(calcular_importe)


# Calcular importe de pedidos
pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]

importes = []

for pedido in pedidos:
    unidades = pedido[0]
    precio = pedido[1]
    descuento = pedido[2]

    importe = calcular_importe(unidades, precio, descuento)
    importes.append(importe)

print("\nImportes:", importes)
print("Máximo:", max(importes))
print("Mínimo:", min(importes))
print("Ordenados:", sorted(importes))
print("Cantidad:", len(importes))
```

## Ejercicio 6 — Normalización de códigos SKU

```python
skus_crudos = [
    "  elc-0012-es ", "HOG-0045-ES", "dep-0003-pt  ",
    "ofi-0021-es", "  ELC-0012-ES", "hog-0099-fr ",
]

# Limpiar datos
skus_limpios = []

for sku in skus_crudos:
    limpio = sku.strip().upper()
    skus_limpios.append(limpio)
    
print("SKU limpios: ", skus_limpios)

# Contar apariciones
aparicionesElc = skus_limpios.count("ELC-0012-ES")
print("\nApariciones de ELC-0012-ES: ", aparicionesElc)

primeraAparicion = skus_limpios.index("OFI-0021-ES")
print("\nPrimera aparición de OFI-0021-ES: ", primeraAparicion)

# Separar SKU
partes_skus = []

for sku in skus_limpios:
    partes = sku.split("-")
    partes_skus.append(partes)

print("\nSKU separados:")
print(partes_skus)

# Filtrar SKU

skus_es = []

for sku in skus_limpios:
    partes = sku.split("-")
    
    if partes[2] == "ES":
        skus_es.append(sku.replace("-", "."))
        
print("\nSKU de España:", skus_es)

# Cadena única
cadena_skus = "| ".join(skus_es)

print("\nCadena final:", cadena_skus)
```

### Ejercicio 8 — Del cálculo con listas al cálculo vectorizado

```python
unidades = [3, 1, 5, 2, 4, 6, 2, 8, 3, 1]
precios  = [59.90, 279.00, 18.50, 149.00, 89.90, 27.50, 219.00, 14.90, 64.00, 239.00]

# Calcular 
print(unidades * precios)
```

Las listas de Python no permiten multiplicación elemento a elemento entre dos listas; para realizar esta operación se necesitan arrays NumPy. 

```python
# numpy arrays
import numpy as np

np_unidades = np.array(unidades)
np_precios = np.array(precios)

print("Unidades:")
print("dtype:", np_unidades.dtype)
print("shape:", np_unidades.shape)

print("\nPrecios:")
print("dtype:", np_precios.dtype)
print("shape:", np_precios.shape)

# Multiplicación elemento por elemento
importes = np_unidades * np_precios

print("Importes: ", importes)
print("Total: ", importes.sum(), "€")

# recargo 
importes_recargo = importes * 1.04

print("Importes con recargo:", importes_recargo)

# array mixto
mixto = np.array([1, "dos", 3.0, True])

print(mixto)
print("dtype:", mixto.dtype)
```

```python
# bucle con NumPy
def calcular_con_for():
    resultado = []

    for i in range(len(unidades)):
        resultado.append(unidades[i] * precios[i])

    return resultado
```

```python
# con %timeit
%timeit calcular_con_for()
```

```python
import numpy as np

# Filas: T1, T2, T3  |  Columnas: Norte, Sur, Este, Oeste, Centro
ventas_2d = np.array([
    [15200.50, 11800.25, 12950.00,  9870.40, 13400.10],
    [16840.75, 12310.60, 13480.90, 11020.35, 14100.80],
    [12535.72, 10663.06, 11221.59, 12673.08, 12334.38],
])

regiones = ["Norte", "Sur", "Este", "Oeste", "Centro"]

print("Shape:", ventas_2d.shape)
print("Ndim:", ventas_2d.ndim)
print("Size:", ventas_2d.size)

# Extraer filas y columnas
segundo_trimestre = ventas_2d[1]

centro = ventas_2d[:,4]

print("\nT2:", segundo_trimestre)
print("Centro:", centro)

# T3 Este
print(ventas_2d[2, 2])

# Substraer con slicing
submatriz = ventas_2d[0:2, 0:3]
print("\nSubmatriz:")
print(submatriz)

# Totales por columna o filas
total_trimestre = ventas_2d.sum(axis=1)
total_region = ventas_2d.sum(axis=0)

print("\nTotal por trimestre:", total_trimestre)
print("Total por región:", total_region)

# Previsión de crecimiento
prevision_t3 = ventas_2d[2] * 1.06
ventas_prevision = np.vstack([ventas_2d, prevision_t3])

print("\nMatriz con previsión:")
print(ventas_prevision)

# Región con mayor facturación
indice_mayor = np.argmax(total_region)
region_mayor = regiones[indice_mayor]

print("\nRegión con mayor facturación:", region_mayor)
```

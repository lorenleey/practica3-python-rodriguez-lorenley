## Ejercicio 1 — Catálogo de productos como lista

```python
catalogo = [
    "Auriculares BT", 59.90,
    "Smartwatch S2", 149.00,
    "Tablet 10", 219.00,
    "Cargador USB-C", 18.50,
    "Cafetera Expres", 189.00,
    "Robot Aspirador", 279.00,
]

# 1. Tomando en cuenta que la lista tiene nombe y precio, asumimos que la mitad de elementos serán los productos
total_elementos = len(catalogo)
total_productos = total_elementos // 2

print("elementos:",total_elementos)
print("productos:",total_productos)

# slicing considerando que los nombres son los números impares, y precios los pares.
nombres = catalogo[::2]
precios = catalogo[1::2]

print("nombres:", nombres)
print("precios:", precios)

# Buscamos "Robot aspirador"
posicion_robot = catalogo.index("Robot Aspirador")
precio_robot = catalogo[posicion_robot + 1]

print("precio robot:", precio_robot)

#Tres ultimos nombres usando indíce negativo
ultimos_tres_nombres = nombres[-3::]
print(ultimos_tres_nombres)

# Precio medio de productos redondeando a dos deciamles unicamente
precio_medio = round(sum(precios) / len(precios),2)
print("precio medio:", precio_medio)
```

## Ejercicio 2 — Inventario multialmacén con listas anidadas

```python
# Declaración de lista con sublistas
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

## Nombre del producto de la cuarta fila (doble indexación)
print("Producto de la cuarta fila:",inventario[3][1])

# Nueva lista donde cada elemento sea [producto, unidades * coste_unitario]
valor_stock = []

for fila in inventario:
    producto = fila[1]
    unidades = fila[2]
    coste_unitario = fila[3]

    valor = round((unidades*coste_unitario),2)
    valor_stock.append([producto, valor])

print("Valor del stock:")
print(valor_stock)

# Calcular la suma de valores del stock de TODOS los productos
valor_total = 0 

for fila in valor_stock:
    valor_total+= fila[1]

print("Valor total del stock:", valor_total)
# Slicing de sublista con solo filas de ALM-SUR 

alm_sur = inventario[2:4]
print("Inventario ALM-SUR:",alm_sur)

valor_sur = 0

for fila in alm_sur:
    valor_sur += round((fila[2]*fila[3]),2)

print("Valor total del stock de ALM-SUR:", valor_sur)
```

#### Información
El problema cuando los dos almacenes no estan contiguos es que si se desea unificar sus datos habría que hacer dos o más sublistas. Eso explica porque slicing no es una estrategia robusta de filtrado, ya que depende de la secuencia y orden de los elementos dentro de la lista. 

```python
precios_originales = [59.90, 149.00, 219.00, 18.50, 189.00, 279.00] 

# Precios rebajados con elementos modificados
precios_rebajados = precios_originales
precios_rebajados[0] = 49.90

print(precios_originales)
print(precios_rebajados)
```

```python
# Copia de una lista con list() y slicing [:]
precios_originales = [59.90, 149.00, 219.00, 18.50, 189.00, 279.00] 
precios_rebajados = list(precios_originales)
precios_rebajados_slc = precios_originales[:]

precios_rebajados[0] = 13.5
precios_rebajados_slc[2] = 100.90

print("Precios originales:",precios_originales)
print("Precios rebajados con list:",precios_rebajados)
print("Precios rebajados slicing:",precios_rebajados_slc)
```

```python
# Aplicar rebaja 10% a todos los elementos
precios_rebajas_10 = list(precios_originales)

for i in range(len(precios_rebajas_10)):
    precios_rebajas_10[i] = round((precios_rebajas_10[i] * 0.90),2)

# Añadir dos productos nuevos
precios_rebajas_10.extend([280.5, 345.2])
precios_rebajas_10.insert(2,560)

# Eliminar producto con del y pop()
del precios_rebajas_10[4]
descartado= precios_rebajas_10.pop()

print("Luego de eliminar y modificar:")
print(precios_rebajas_10)
print("Valor descartado:", descartado)

# Comprobar id 
print("\nid original:", id(precios_originales))
print("id precios list:", id(precios_rebajados))
print("id precios slicing:", id(precios_rebajados_slc))
print("id precios rebajados 10%:", id(precios_rebajas_10))

```

## Ejercicio 4 — Cola de pedidos pendientes

```python
cola = ["PED-10021", "PED-10007", "PED-10044", "PED-10012", "PED-10033", "PED-10008"]

total_inicial = len(cola)
# Invertir cola y guardarlo en una lista 
cola_reversa = cola[::-1] 
print(cola_reversa)

# Ordenar cola con sorted() sin alterar la original
cola_ordenada = sorted(cola)
print("Cola ordenada:",cola_ordenada)
print("Cola original:",cola)

# Ordenar cola con .sort(reverse=True)
cola.sort(reverse=True)
print("Cola ordenada a la reversa:", cola)

# Atender los dos primeros pedidos con .pop(0)
atendidos = []
atendidos.append(cola.pop(0))
atendidos.append(cola.pop(0))

print(atendidos)

# Insertar pedidos urgentes 
cola.insert(0, "PED-99999")

# Informe final

num_atendidos = len(atendidos)
num_pendientes = len(cola)
porcentaje_avance = round((len(atendidos) / total_inicial *100),1)

print("\n Número de pedidos atendidos:", num_atendidos)
print("Número de pedidos pendientes:", num_pendientes)
print("Pedidos pendientes:", cola)
print("Porcentaje de avance en pedidos:", porcentaje_avance) 

```
